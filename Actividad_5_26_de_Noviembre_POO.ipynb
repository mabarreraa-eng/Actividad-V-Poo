```python

# ==========================================
# Ejercicio 8.4
# ==========================================
import ipywidgets as widgets
from IPython.display import display, HTML
from dataclasses import dataclass
from enum import Enum
from google.colab import files

# =========================
# TIPOS DE EMPLEADO
# =========================

class TipoCargo(Enum):
    DIRECTIVO = "Directivo"
    ESTRATEGICO = "Estratégico"
    OPERATIVO = "Operativo"


class TipoGenero(Enum):
    MASCULINO = "Masculino"
    FEMENINO = "Femenino"


# =========================
# CLASE EMPLEADO
# =========================

@dataclass
class Empleado:
    nombre: str
    apellidos: str
    cargo: TipoCargo
    genero: TipoGenero
    salario_dia: float
    dias_trabajados: int
    otros_ingresos: float
    pagos_salud: float
    aportes_pensiones: float

    def calcular_nomina(self):
        salario = self.salario_dia * self.dias_trabajados
        salario += self.otros_ingresos
        salario -= self.pagos_salud
        salario -= self.aportes_pensiones
        return salario


# =========================
# CLASE LISTA DE EMPLEADOS
# =========================

class ListaEmpleados:

    def __init__(self):
        self.empleados = []

    def agregar(self, empleado):
        self.empleados.append(empleado)

    def calcular_total_nomina(self):
        return sum(
            empleado.calcular_nomina()
            for empleado in self.empleados
        )

    def borrar_todos(self):
        self.empleados.clear()

    def guardar_archivo(self):

        with open("Nomina.txt", "w", encoding="utf-8") as archivo:

            archivo.write("NÓMINA DE EMPLEADOS\n")
            archivo.write("=" * 45 + "\n\n")

            for empleado in self.empleados:

                archivo.write(f"Nombre: {empleado.nombre}\n")
                archivo.write(f"Apellidos: {empleado.apellidos}\n")
                archivo.write(f"Cargo: {empleado.cargo.value}\n")
                archivo.write(f"Género: {empleado.genero.value}\n")

                archivo.write(
                    f"Salario por día: ${empleado.salario_dia:,.2f}\n"
                )

                archivo.write(
                    f"Días trabajados al mes: {empleado.dias_trabajados}\n"
                )

                archivo.write(
                    f"Otros ingresos: ${empleado.otros_ingresos:,.2f}\n"
                )

                archivo.write(
                    f"Pagos por salud: ${empleado.pagos_salud:,.2f}\n"
                )

                archivo.write(
                    f"Aportes pensiones: ${empleado.aportes_pensiones:,.2f}\n"
                )

                archivo.write(
                    f"Sueldo mensual: ${empleado.calcular_nomina():,.2f}\n"
                )

                archivo.write("-" * 45 + "\n")

            archivo.write(
                f"\nTOTAL NÓMINA: ${self.calcular_total_nomina():,.2f}"
            )


# =========================
# VENTANA AGREGAR EMPLEADO
# =========================

class VentanaAgregarEmpleado:

    def __init__(self, lista_empleados, actualizar):

        self.lista_empleados = lista_empleados
        self.actualizar = actualizar

        self.estilo_campo = widgets.Layout(
            width="100%",
            height="35px"
        )

        # Campos de información

        self.nombre = widgets.Text(
            layout=self.estilo_campo
        )

        self.apellidos = widgets.Text(
            layout=self.estilo_campo
        )

        self.cargo = widgets.Dropdown(
            options=[c.value for c in TipoCargo],
            value="Directivo",
            layout=self.estilo_campo
        )

        self.genero = widgets.RadioButtons(
            options=["Masculino", "Femenino"],
            value="Masculino",
            layout=widgets.Layout(width="100%")
        )

        self.salario = widgets.FloatText(
            value=0,
            layout=self.estilo_campo
        )

        self.dias = widgets.BoundedIntText(
            value=0,
            min=0,
            max=31,
            layout=self.estilo_campo
        )

        self.otros = widgets.FloatText(
            value=0,
            layout=self.estilo_campo
        )

        self.salud = widgets.FloatText(
            value=0,
            layout=self.estilo_campo
        )

        self.pensiones = widgets.FloatText(
            value=0,
            layout=self.estilo_campo
        )

        self.salida = widgets.Output()

        # Botones

        self.boton_agregar = self.crear_boton(
            "Agregar", "#8056A3"
        )

        self.boton_limpiar = self.crear_boton(
            "Limpiar", "#B99AD5"
        )

        self.boton_borrar = self.crear_boton(
            "Borrar", "#593477"
        )

        self.boton_agregar.on_click(self.agregar)
        self.boton_limpiar.on_click(self.limpiar)
        self.boton_borrar.on_click(self.borrar)

        self.formulario = self.crear_formulario()

    # Crear botones

    def crear_boton(self, texto, color):

        return widgets.Button(
            description=texto,
            layout=widgets.Layout(
                width="100%",
                height="42px"
            ),
            style={
                "button_color": color,
                "font_weight": "bold"
            }
        )

    # Crear filas del formulario

    def crear_fila(self, etiqueta, campo):

        texto = widgets.HTML(
            value=f"""
            <div style="
                width:100%;
                font-size:14px;
                padding-top:8px;
                white-space:normal;
            ">
                {etiqueta}
            </div>
            """,
            layout=widgets.Layout(
                width="230px",
                min_width="230px"
            )
        )

        return widgets.HBox(
            [texto, campo],
            layout=widgets.Layout(
                width="100%",
                align_items="center"
            )
        )

    # Construir formulario

    def crear_formulario(self):

        filas = [
            self.crear_fila("Nombre:", self.nombre),
            self.crear_fila("Apellidos:", self.apellidos),
            self.crear_fila("Cargo:", self.cargo),
            self.crear_fila("Género:", self.genero),
            self.crear_fila("Salario por día:", self.salario),
            self.crear_fila("Días trabajados al mes:", self.dias),
            self.crear_fila("Otros ingresos:", self.otros),
            self.crear_fila("Pagos por salud:", self.salud),
            self.crear_fila("Aportes pensiones:", self.pensiones)
        ]

        botones = widgets.HBox(
            [
                self.boton_agregar,
                self.boton_limpiar,
                self.boton_borrar
            ],
            layout=widgets.Layout(
                width="100%",
                gap="8px"
            )
        )

        return widgets.VBox(
            filas + [botones, self.salida],
            layout=widgets.Layout(
                width="100%",
                gap="7px",
                padding="10px"
            )
        )

    # Limpiar únicamente el formulario

    def limpiar_campos(self):

        self.nombre.value = ""
        self.apellidos.value = ""

        self.cargo.value = "Directivo"
        self.genero.value = "Masculino"

        self.salario.value = 0
        self.dias.value = 0
        self.otros.value = 0
        self.salud.value = 0
        self.pensiones.value = 0

    # Acción del botón Limpiar

    def limpiar(self, boton):

        self.limpiar_campos()
        self.salida.clear_output()

        with self.salida:
            print("Formulario limpiado correctamente.")

    # Acción del botón Agregar

    def agregar(self, boton):

        with self.salida:
            self.salida.clear_output()

            if not self.nombre.value.strip():
                print("Ingrese el nombre del empleado.")
                return

            if not self.apellidos.value.strip():
                print("Ingrese los apellidos del empleado.")
                return

            if self.dias.value < 1:
                print("Los días trabajados deben estar entre 1 y 31.")
                return

            valores = [
                self.salario.value,
                self.otros.value,
                self.salud.value,
                self.pensiones.value
            ]

            if any(valor < 0 for valor in valores):
                print("Los valores numéricos no pueden ser negativos.")
                return

            cargo_seleccionado = next(
                c for c in TipoCargo
                if c.value == self.cargo.value
            )

            genero_seleccionado = next(
                g for g in TipoGenero
                if g.value == self.genero.value
            )

            empleado = Empleado(
                nombre=self.nombre.value.strip(),
                apellidos=self.apellidos.value.strip(),
                cargo=cargo_seleccionado,
                genero=genero_seleccionado,
                salario_dia=self.salario.value,
                dias_trabajados=self.dias.value,
                otros_ingresos=self.otros.value,
                pagos_salud=self.salud.value,
                aportes_pensiones=self.pensiones.value
            )

            self.lista_empleados.agregar(empleado)

            print("Empleado agregado correctamente.")

            self.actualizar()

            # Limpiar el formulario sin borrar la lista

            self.limpiar_campos()

    # Acción del botón Borrar

    def borrar(self, boton):

        # Eliminar todos los empleados

        self.lista_empleados.borrar_todos()

        # Limpiar el formulario

        self.limpiar_campos()

        # Limpiar resultados anteriores

        self.salida.clear_output()
        self.actualizar()

        with self.salida:
            print("Todos los empleados fueron eliminados.")


# =========================
# VENTANA NÓMINA
# =========================

class VentanaNomina:

    def __init__(self, lista_empleados):
        self.lista_empleados = lista_empleados

    def mostrar(self):

        empleados = self.lista_empleados.empleados

        if not empleados:
            display(
                HTML("<p>No hay empleados registrados.</p>")
            )
            return

        filas = ""

        for empleado in empleados:

            filas += f"""
            <tr>
                <td style="padding:10px;border:1px solid #aaa;">
                    {empleado.nombre}
                </td>

                <td style="padding:10px;border:1px solid #aaa;">
                    {empleado.apellidos}
                </td>

                <td style="padding:10px;border:1px solid #aaa;">
                    ${empleado.calcular_nomina():,.2f}
                </td>
            </tr>
            """

        total = self.lista_empleados.calcular_total_nomina()

        tabla = f"""
        <table style="
            border-collapse:collapse;
            width:100%;
            text-align:left;
        ">

            <tr>
                <th style="padding:10px;border:1px solid #aaa;">
                    Nombre
                </th>

                <th style="padding:10px;border:1px solid #aaa;">
                    Apellidos
                </th>

                <th style="padding:10px;border:1px solid #aaa;">
                    Sueldo mensual
                </th>
            </tr>

            {filas}

        </table>

        <h3>Total nómina: ${total:,.2f}</h3>
        """

        display(HTML(tabla))


# =========================
# VENTANA PRINCIPAL
# =========================

class VentanaPrincipal:

    def __init__(self):

        self.lista_empleados = ListaEmpleados()

        self.ventana_nomina = VentanaNomina(
            self.lista_empleados
        )

        self.resultados = widgets.Output()

        self.ventana_agregar = VentanaAgregarEmpleado(
            self.lista_empleados,
            self.actualizar
        )

        self.boton_agregar = self.crear_boton(
            "Agregar empleado", "#8056A3"
        )

        self.boton_calcular = self.crear_boton(
            "Calcular nómina", "#B99AD5"
        )

        self.boton_guardar = self.crear_boton(
            "Guardar archivo", "#593477"
        )

        self.boton_agregar.on_click(
            self.ventana_agregar.agregar
        )

        self.boton_calcular.on_click(
            self.calcular_nomina
        )

        self.boton_guardar.on_click(
            self.guardar_archivo
        )

    # Crear botones principales

    def crear_boton(self, texto, color):

        return widgets.Button(
            description=texto,
            layout=widgets.Layout(
                width="100%",
                height="42px"
            ),
            style={
                "button_color": color,
                "font_weight": "bold"
            }
        )

    # Actualizar resultados

    def actualizar(self):

        self.resultados.clear_output()

    # Calcular nómina

    def calcular_nomina(self, boton):

        with self.resultados:
            self.resultados.clear_output()
            self.ventana_nomina.mostrar()

    # Guardar archivo

    def guardar_archivo(self, boton):

        with self.resultados:
            self.resultados.clear_output()

            if not self.lista_empleados.empleados:
                print("No hay empleados para guardar.")
                return

            self.lista_empleados.guardar_archivo()

            print("Archivo guardado correctamente.")

            files.download("Nomina.txt")

    # Mostrar ventana principal

    def mostrar(self):

        display(HTML("""
        <h2 style="color:white;">Nómina</h2>

        <p style="color:white;">
            Gestión de empleados y cálculo de nómina
        </p>
        """))

        menu = widgets.HTML(
            "<h4 style='margin-bottom:5px;'>Menú</h4>"
        )

        botones = widgets.VBox(
            [
                self.boton_agregar,
                self.boton_calcular,
                self.boton_guardar
            ],
            layout=widgets.Layout(
                width="250px",
                gap="3px"
            )
        )

        display(menu)
        display(botones)

        display(HTML("<h3>Agregar Empleado</h3>"))

        display(self.ventana_agregar.formulario)

        display(self.resultados)


# =========================
# EJECUTAR PROGRAMA
# =========================

programa = VentanaPrincipal()
programa.mostrar()
```

```python

# ==========================================
# Ejercicio 8.5
# ==========================================
import ipywidgets as widgets
from IPython.display import display, HTML
from dataclasses import dataclass
from datetime import datetime


# ==========================================
# CLASE HUESPED
# ==========================================

@dataclass
class Huesped:
    nombre: str
    apellidos: str
    documento: str
    fecha_ingreso: datetime


# ==========================================
# CLASE HABITACION
# ==========================================

class Habitacion:

    def __init__(self, numero, precio_dia):
        self.numero = numero
        self.precio_dia = precio_dia
        self.disponible = True
        self.huesped = None

    def registrar_huesped(self, huesped):
        self.huesped = huesped
        self.disponible = False

    def liberar(self):
        self.huesped = None
        self.disponible = True


# ==========================================
# CLASE HOTEL
# ==========================================

class Hotel:

    def __init__(self):

        self.habitaciones = []

        for numero in range(1, 11):

            if numero <= 5:
                precio = 120000
            else:
                precio = 160000

            self.habitaciones.append(
                Habitacion(numero, precio)
            )

    def buscar_habitacion(self, numero):

        if 1 <= numero <= 10:
            return self.habitaciones[numero - 1]

        return None

    def buscar_habitacion_ocupada(self, numero):

        habitacion = self.buscar_habitacion(numero)

        return (
            habitacion is not None
            and not habitacion.disponible
        )

    def obtener_habitaciones_ocupadas(self):

        return [
            h for h in self.habitaciones
            if not h.disponible
        ]


# ==========================================
# VENTANA PRINCIPAL
# ==========================================

class VentanaPrincipal:

    def __init__(self):

        self.hotel = Hotel()

        # Panel único para cambiar de pantalla
        self.panel = widgets.Output()

        self.boton_consultar = self.crear_boton(
            "Consultar habitaciones", "#8056A3"
        )

        self.boton_salida = self.crear_boton(
            "Salida de huéspedes", "#B99AD5"
        )

        self.boton_consultar.on_click(
            self.consultar_habitaciones
        )

        self.boton_salida.on_click(
            self.mostrar_salida
        )

    # --------------------------------------
    # CREAR BOTONES
    # --------------------------------------

    def crear_boton(self, texto, color):

        return widgets.Button(
            description=texto,
            layout=widgets.Layout(
                width="100%",
                height="42px"
            ),
            style={
                "button_color": color,
                "font_weight": "bold"
            }
        )

    # --------------------------------------
    # CAMBIAR PANTALLA
    # --------------------------------------

    def mostrar_pantalla(self, contenido):

        with self.panel:
            self.panel.clear_output(wait=True)
            display(contenido)

    # --------------------------------------
    # MENSAJES
    # --------------------------------------

    def mensaje(self, texto, color="#333333"):

        display(HTML(
            f"<p style='color:{color};font-weight:bold;'>{texto}</p>"
        ))

    # --------------------------------------
    # MENU PRINCIPAL
    # --------------------------------------

    def mostrar(self):

        titulo = widgets.HTML(
            "<h2 style='color:white;'>Hotel</h2>"
        )

        menu = widgets.HTML("<h4>Menú</h4>")

        botones = widgets.VBox(
            [
                self.boton_consultar,
                self.boton_salida
            ],
            layout=widgets.Layout(
                width="280px",
                gap="3px"
            )
        )

        interfaz = widgets.VBox([
            titulo,
            menu,
            botones,
            self.panel
        ])

        display(interfaz)

    def volver_menu(self, boton=None):
        self.mostrar_pantalla(
            widgets.HTML("<h3>Seleccione una opción del menú.</h3>")
        )

    # ======================================
    # CONSULTAR HABITACIONES
    # ======================================

    def consultar_habitaciones(self, boton):

        contenido = widgets.VBox(
            layout=widgets.Layout(
                width="100%",
                gap="12px"
            )
        )

        contenido.children = (
            widgets.HTML("<h3>Habitaciones</h3>"),
        )

        filas = []

        for habitacion in self.hotel.habitaciones:

            if habitacion.disponible:
                estado = "Disponible"
                color = "#238636"
            else:
                estado = "No disponible"
                color = "#d93025"

            tarjeta = widgets.VBox(
                [
                    widgets.HTML(
                        f"<b>Habitación {habitacion.numero}</b>"
                    ),
                    widgets.HTML(
                        f"<span style='color:{color};'>{estado}</span>"
                    ),
                    widgets.HTML(
                        f"${habitacion.precio_dia:,.0f} por día"
                    )
                ],
                layout=widgets.Layout(
                    width="140px",
                    border="1px solid #aaa",
                    padding="10px",
                    gap="5px"
                )
            )

            filas.append(tarjeta)

        fila1 = widgets.HBox(
            filas[:5],
            layout=widgets.Layout(
                width="100%",
                gap="8px",
                justify_content="space-between"
            )
        )

        fila2 = widgets.HBox(
            filas[5:],
            layout=widgets.Layout(
                width="100%",
                gap="8px",
                justify_content="space-between"
            )
        )

        selector = widgets.BoundedIntText(
            value=1,
            min=1,
            max=10,
            description="Habitación a reservar:",
            style={"description_width": "initial"},
            layout=widgets.Layout(width="300px")
        )

        aceptar = self.crear_boton(
            "Aceptar", "#8056A3"
        )

        volver = self.crear_boton(
            "Volver al menú", "#B99AD5"
        )

        mensaje = widgets.Output()

        def seleccionar(b):

            habitacion = self.hotel.buscar_habitacion(
                selector.value
            )

            if not habitacion.disponible:

                with mensaje:
                    mensaje.clear_output()
                    print("La habitación está ocupada.")

                return

            self.mostrar_ingreso(habitacion)

        aceptar.on_click(seleccionar)
        volver.on_click(self.volver_menu)

        contenido.children = (
            widgets.HTML("<h3>Habitaciones</h3>"),
            fila1,
            fila2,
            selector,
            widgets.HBox([aceptar, volver]),
            mensaje
        )

        self.mostrar_pantalla(contenido)

    # ======================================
    # INGRESO DEL HUESPED
    # ======================================

    def mostrar_ingreso(self, habitacion):

        titulo = widgets.HTML(
            f"<h3>Ingreso - Habitación {habitacion.numero}</h3>"
        )

        fecha = widgets.Text(
            description="Fecha (aaaa-mm-dd):",
            placeholder="2026-11-24",
            style={"description_width": "initial"},
            layout=widgets.Layout(width="400px")
        )

        nombre = widgets.Text(
            description="Nombre:",
            style={"description_width": "initial"},
            layout=widgets.Layout(width="400px")
        )

        apellidos = widgets.Text(
            description="Apellidos:",
            style={"description_width": "initial"},
            layout=widgets.Layout(width="400px")
        )

        documento = widgets.Text(
            description="Doc. Identidad:",
            style={"description_width": "initial"},
            layout=widgets.Layout(width="400px")
        )

        aceptar = self.crear_boton(
            "Aceptar", "#8056A3"
        )

        cancelar = self.crear_boton(
            "Cancelar", "#B99AD5"
        )

        resultado = widgets.Output()

        def registrar(b):

            if not all([
                fecha.value.strip(),
                nombre.value.strip(),
                apellidos.value.strip(),
                documento.value.strip()
            ]):

                with resultado:
                    resultado.clear_output()
                    print("Todos los campos son obligatorios.")

                return

            try:

                fecha_ingreso = datetime.strptime(
                    fecha.value.strip(),
                    "%Y-%m-%d"
                )

            except ValueError:

                with resultado:
                    resultado.clear_output()
                    print("La fecha no está en el formato solicitado.")

                return

            if not documento.value.strip().isdigit():

                with resultado:
                    resultado.clear_output()
                    print("El documento debe contener solo números.")

                return

            if not habitacion.disponible:

                with resultado:
                    resultado.clear_output()
                    print("La habitación está ocupada.")

                return

            huesped = Huesped(
                nombre.value.strip(),
                apellidos.value.strip(),
                documento.value.strip(),
                fecha_ingreso
            )

            habitacion.registrar_huesped(huesped)

            self.mostrar_pantalla(
                widgets.VBox([
                    widgets.HTML("<h3>Mensaje</h3>"),
                    widgets.HTML(
                        "<p>El huésped ha sido registrado.</p>"
                    ),
                    self.crear_boton(
                        "Volver a habitaciones", "#8056A3"
                    )
                ])
            )

            # Conectar el botón de regreso
            self.panel.clear_output(wait=True)

            boton_volver = self.crear_boton(
                "Volver a habitaciones", "#8056A3"
            )

            boton_volver.on_click(
                self.consultar_habitaciones
            )

            self.mostrar_pantalla(
                widgets.VBox([
                    widgets.HTML("<h3>Mensaje</h3>"),
                    widgets.HTML(
                        "<p>El huésped ha sido registrado.</p>"
                    ),
                    boton_volver
                ])
            )

        aceptar.on_click(registrar)
        cancelar.on_click(self.consultar_habitaciones)

        formulario = widgets.VBox([
            titulo,
            fecha,
            widgets.HTML("<b>Huésped</b>"),
            nombre,
            apellidos,
            documento,
            widgets.HBox([aceptar, cancelar]),
            resultado
        ], layout=widgets.Layout(
            width="500px",
            gap="10px"
        ))

        self.mostrar_pantalla(formulario)

    # ======================================
    # SALIDA DE HUESPEDES
    # ======================================

    def mostrar_salida(self, boton):

        ocupadas = self.hotel.obtener_habitaciones_ocupadas()

        if not ocupadas:

            self.mostrar_pantalla(
                widgets.VBox([
                    widgets.HTML("<h3>Salida de huéspedes</h3>"),
                    widgets.HTML(
                        "<p>No hay habitaciones ocupadas.</p>"
                    ),
                    self.crear_boton(
                        "Volver al menú", "#B99AD5"
                    )
                ])
            )

            return

        opciones = [
            (f"Habitación {h.numero}", h.numero)
            for h in ocupadas
        ]

        selector = widgets.Dropdown(
            options=opciones,
            description="Número de habitación:",
            style={"description_width": "initial"},
            layout=widgets.Layout(width="320px")
        )

        aceptar = self.crear_boton(
            "Aceptar", "#8056A3"
        )

        cancelar = self.crear_boton(
            "Cancelar", "#B99AD5"
        )

        resultado = widgets.Output()

        def seleccionar(b):

            habitacion = self.hotel.buscar_habitacion(
                selector.value
            )

            if habitacion is None or habitacion.disponible:

                with resultado:
                    resultado.clear_output()
                    print("La habitación ingresada no ha sido ocupada.")

                return

            self.mostrar_fecha_salida(habitacion)

        aceptar.on_click(seleccionar)
        cancelar.on_click(self.volver_menu)

        formulario = widgets.VBox([
            widgets.HTML("<h3>Salida de huéspedes</h3>"),
            widgets.HTML("<p>Ingrese número de habitación</p>"),
            selector,
            widgets.HBox([aceptar, cancelar]),
            resultado
        ], layout=widgets.Layout(
            width="450px",
            gap="10px"
        ))

        self.mostrar_pantalla(formulario)

    # ======================================
    # FECHA Y CALCULO DE SALIDA
    # ======================================

    def mostrar_fecha_salida(self, habitacion):

        huesped = habitacion.huesped

        fecha_ingreso = huesped.fecha_ingreso

        fecha_salida = widgets.Text(
            description="Fecha de salida (aaaa-mm-dd):",
            placeholder="2026-11-26",
            style={"description_width": "initial"},
            layout=widgets.Layout(width="450px")
        )

        calcular = self.crear_boton(
            "Calcular", "#8056A3"
        )

        registrar = self.crear_boton(
            "Registrar salida", "#593477"
        )

        registrar.disabled = True

        resultado = widgets.Output()

        datos = {
            "dias": None,
            "total": None,
            "fecha": None
        }

        def calcular_pago(b):

            registrar.disabled = True

            try:

                fecha = datetime.strptime(
                    fecha_salida.value.strip(),
                    "%Y-%m-%d"
                )

            except ValueError:

                with resultado:
                    resultado.clear_output()
                    print("La fecha no está en el formato solicitado.")

                return

            if fecha <= fecha_ingreso:

                with resultado:
                    resultado.clear_output()
                    print(
                        "La fecha de salida debe ser mayor "
                        "a la fecha de ingreso."
                    )

                return

            dias = (fecha - fecha_ingreso).days
            total = dias * habitacion.precio_dia

            datos["dias"] = dias
            datos["total"] = total
            datos["fecha"] = fecha

            registrar.disabled = False

            with resultado:

                resultado.clear_output()

                print(f"Cantidad de días: {dias}")
                print(f"Total: ${total:,.0f}")

        def registrar_salida(b):

            if datos["dias"] is None:
                return

            habitacion.liberar()

            registrar.disabled = True
            calcular.disabled = True

            with resultado:

                resultado.clear_output()

                print("Se ha registrado la salida del huésped.")

            boton_volver = self.crear_boton(
                "Volver al menú", "#B99AD5"
            )

            boton_volver.on_click(self.volver_menu)

            with self.panel:
                display(boton_volver)

        calcular.on_click(calcular_pago)
        registrar.on_click(registrar_salida)

        formulario = widgets.VBox([
            widgets.HTML(
                f"<h3>Salida de huéspedes</h3>"
                f"<b>Habitación: {habitacion.numero}</b>"
            ),
            widgets.HTML(
                f"Fecha de ingreso: {fecha_ingreso.strftime('%Y-%m-%d')}"
            ),
            fecha_salida,
            calcular,
            resultado,
            registrar
        ], layout=widgets.Layout(
            width="480px",
            gap="10px"
        ))

        self.mostrar_pantalla(formulario)


# ==========================================
# EJECUTAR PROGRAMA
# ==========================================

programa = VentanaPrincipal()
programa.mostrar()
```
